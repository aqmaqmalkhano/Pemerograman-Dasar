```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
aqm = pd.read_csv('/content/products.csv')
#import file csv
```

```python
aqm.describe()
```

```python
# Ringkasan statistik
aqm.describe()

# Rata-rata harga
rata_rata_price = aqm['price'].mean()
print("Rata-rata harga:", rata_rata_price)

# Modus jumlah quantity
modus_qty = aqm['quantity'].mode()
print("Modus size:", modus_qty)
```

```python
aqm['quantity'].hist()
plt.title("frekuensi jumlah Quantity")
plt.xlabel("Quantity")
plt.ylabel("frekuensi")
plt.show()
```

```python
# Korelasi
corr_matrix = aqm[['quantity', 'price']].corr()
print("Korelasi:", corr_matrix)

# Kovarian
cov_matrix = aqm[['quantity', 'price']].cov()
print("Kovarian:", cov_matrix)
```

```python
KOLOM_GROUPING = 'product_type'

group_mean_data = aqm.groupby(KOLOM_GROUPING).mean(numeric_only=True)
print(group_mean_data)

group_agg_data = aqm.groupby(KOLOM_GROUPING).agg({
    'price': ['mean', 'min', 'max'],
    'quantity': ['mean', 'min', 'max']
})
print(group_agg_data)
```
