```python
# Input
n = int(input("Masukkan bilangan untuk dihitung faktorialnya: "))

# Hitung faktorial dengan FOR
faktorial_for = 1
for i in range(1, n + 1):
    faktorial_for *= i

print(f"Faktorial {n} (menggunakan FOR) = {faktorial_for}")
```

```python
# Hitung faktorial dengan WHILE
faktorial_while = 1
counter = 1
while counter <= n:
    faktorial_while *= counter
    counter += 1

print(f"Faktorial {n} (menggunakan WHILE) = {faktorial_while}")
```
